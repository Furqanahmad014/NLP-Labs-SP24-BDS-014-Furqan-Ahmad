## Task 1: Regex Fundamentals (Character Classes & Quantifiers)

```python
import re

sample_text = (
    "In 2023, Ahmed bought 3 Laptops and 2 Phones for $1500. "
    "The event took place in 1999 and will repeat in 2025. "
    "Cat, Dog, Elephant and Ant were seen near the Old Zoo Gate #42."
)
print(sample_text)
```

**a) Extract all digits from the text**

```python
digits = re.findall(r'\d', sample_text)
print(digits)
```

**b) Extract all words that start with a capital letter**

```python
capitalized_words = re.findall(r'\b[A-Z][a-zA-Z]*\b', sample_text)
print(capitalized_words)
```

**c) Extract all sequences of exactly 4 digits (e.g., years)**

```python
four_digit_sequences = re.findall(r'\b\d{4}\b', sample_text)
print(four_digit_sequences)
```

**d) Extract all words that are between 3 to 6 characters long**

```python
words_3_to_6 = re.findall(r'\b[a-zA-Z]{3,6}\b', sample_text)
print(words_3_to_6)
```

## Task 2: Pattern-Based Extraction (Real World Data)

```python
real_world_text = (
    "Please contact us at support@nlp.com or sales.team@company.pk for queries. "
    "You can also reach Ahmed at ahmed_khan99@gmail.com. "
    "Call our helpline at 0301-1234567 or 0333-9876543. "
    "Your order #ORD1023 was placed on 15/08/2024 and delivered on 20/08/2024. "
    "The discount applied was 45% off, while the tax was 12.5% on the remaining amount."
)
print(real_world_text)
```

**a) Extract all email addresses using `re.findall()`**

```python
emails = re.findall(r'[\w.+-]+@[\w-]+\.[\w.-]+', real_world_text)
print(emails)
```

**b) Extract all phone numbers (format: 03XX-XXXXXXX)**

```python
phone_numbers = re.findall(r'03\d{2}-\d{7}', real_world_text)
print(phone_numbers)
```

**c) Extract all dates in DD/MM/YYYY format**

```python
dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', real_world_text)
print(dates)
```

**d) Extract all percentage values (e.g., 45%, 12.5%)**

```python
percentages = re.findall(r'\b\d+(?:\.\d+)?%', real_world_text)
print(percentages)
```

## Task 3: Word Tokenization (NLTK vs spaCy)

```python
paragraph = (
    "I don't think it's going to rain today, but Dr. Ahmed said he'll bring an umbrella anyway. "
    "She's already left for the office, and she can't wait to see the new project. "
    "Isn't it amazing how technology's changed the world?"
)
print(paragraph)
```

**a) Tokenize the paragraph using NLTK's `word_tokenize()`**

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

nltk_tokens = word_tokenize(paragraph)
print(nltk_tokens)
```

**b) Tokenize the same paragraph using a blank spaCy pipeline**

```python
import spacy

nlp_blank = spacy.blank("en")
doc_blank = nlp_blank(paragraph)
spacy_tokens = [token.text for token in doc_blank]
print(spacy_tokens)
```

**c) Comparison of the two outputs (at least 2 differences)**

```python
# Observations:
# 1. NLTK's word_tokenize splits contractions into two separate tokens
#    (e.g., "do", "n't" / "'s" as its own token), whereas spaCy's blank
#    pipeline (rule-based tokenizer) often keeps contractions like
#    "don't" attached differently or splits them using its own exception
#    rules, giving a slightly different split (e.g., "do" + "n't" vs "don't").
# 2. NLTK tends to separate punctuation more consistently as standalone
#    tokens, while spaCy's tokenizer uses prefix/suffix/infix rules that
#    can treat certain punctuation + word combinations differently
#    (e.g., handling of possessive "'s" or trailing periods after "Dr.").
print("See comment above for the comparison.")
```

**d) Count the total number of tokens produced by each method**

```python
print("NLTK token count:", len(nltk_tokens))
print("spaCy (blank) token count:", len(spacy_tokens))
```

## Task 4: Token Attributes (Categorizing Tokens)

```python
import spacy

nlp = spacy.load("en_core_web_sm")
sample_sentence = "Ahmed bought 2 laptops worth $50 and paid via ahmed@example.com using Rs. 500 cash."
print(sample_sentence)
```

**a) Load the sentence into a spaCy `Doc` object**

```python
doc = nlp(sample_sentence)
print(doc)
```

**b) Loop through each token and print its `is_alpha` value**

```python
for token in doc:
    print(f"{token.text!r:20} is_alpha = {token.is_alpha}")
```

**c) Print each token's `like_num` and `like_email` values**

```python
for token in doc:
    print(f"{token.text!r:20} like_num = {token.like_num!s:6} like_email = {token.like_email}")
```

**d) Identify and list only the tokens where `is_currency` is True**

```python
currency_tokens = [token.text for token in doc if token.is_currency]
print("Currency tokens:", currency_tokens)
```

## Task 5: Sentence Segmentation (Handling Tricky Cases)

**a) Load a trained spaCy pipeline (`en_core_web_sm`)**

```python
# nlp already loaded above with spacy.load("en_core_web_sm")
tricky_paragraph = (
    "Dr. Ahmed visited the clinic early in the morning. "
    "He met Mr. Khan, who works there as e.g. a lab assistant, and discussed the reports. "
    "They finished the meeting before 10 a.m. and left together."
)
print(tricky_paragraph)
```

**b) Use `doc.sents` to split the paragraph into individual sentences**

```python
doc2 = nlp(tricky_paragraph)
sentences = list(doc2.sents)
for i, sent in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sent.text}")
```

**c) Print the total number of sentences detected**

```python
print("Total sentences detected:", len(sentences))
```

**d) Did the abbreviation cause any incorrect sentence break? Observation**

```python
# Observation: The trained en_core_web_sm pipeline correctly handled "Dr." and "Mr."
# without breaking the sentence there, since it recognizes these as common
# abbreviations from its statistical model. However, "e.g." in the middle of a
# sentence can sometimes still be a risk point for rule-based/blank tokenizers;
# here the trained pipeline kept it correctly within the same sentence,
# showing the trained model handles abbreviations better than naive
# period-based splitting would.
print("See comment above for the observation.")
```
